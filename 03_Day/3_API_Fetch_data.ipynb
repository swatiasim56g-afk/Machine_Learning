```python
# Now usnig loop fetching data through API and storing it in a dataframe

import pandas as pd
import requests

response = requests.get('https://api.example.com/data')
data = response.json()
df = pd.DataFrame(data.json().[['key1', 'key2', 'key3']])

for i in range(1, 10):
    response = requests.get(f'https://api.example.com/data?page={i}')
    data = response.json()
    df = pd.concat([df, pd.DataFrame(data.json().[['key1', 'key2', 'key3']])], ignore_index=True)
```

```python
# now convert into csv file

df.to_csv('data.csv', index=False)
```
