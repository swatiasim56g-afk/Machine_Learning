# Feature Scaling 

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import libraries for model training
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

## Upload dataset

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
df.head(5)
```

```python

x_train, x_test, y_train, y_test = train_test_split(df[['Age', 'EstimatedSalary']], df['Purchased'], test_size=0.2, random_state=0)
x_train.shape, x_test.shape
```

## StandartScaler

```python
# Standard Scaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
# now set the scaler to the training data
scaler.fit(x_train)
# transform the training and testing data
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
scaler.mean_
```

```python
train_df = pd.DataFrame(x_train_scaled, columns=['Age', 'EstimatedSalary'])
train_df.head(5)
```

```python
pd.DataFrame(x_train_scaled, columns=['Age', 'EstimatedSalary']).describe().round(1)
```

```python
# trian model on scaled data and unscaled data and compare the accuracy score

model = LogisticRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
# pirnt accuracy score for scaled data and unscaled data
print("Accuracy score for unscaled data: ", accuracy_score(y_test, y_pred))
print("Accuracy score for scaled data: ", accuracy_score(y_test, model.predict(x_test_scaled))) 
```
