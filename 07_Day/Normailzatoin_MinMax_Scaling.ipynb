# MINMAX Scaling ON WINE Data set

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import ml libraries
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
```

```python
# upload wine dataset with specific columns 
pf = pd.read_csv('WineQT.csv', usecols=['fixed acidity', 'volatile acidity', 'free sulfur dioxide'])
pf.head()
```

```python
# make a deskplot 
sns.pairplot(pf)
plt.show()
```

```python
# make a scatter plot 
pf.plot(kind='scatter', x='fixed acidity', y='volatile acidity')


# make a scatter plot 
pf.plot(kind='scatter', x='fixed acidity', y='free sulfur dioxide')
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
X_train, x_test, Y_train, y_test = train_test_split(pf[['fixed acidity']], pf['volatile acidity'], test_size=0.2, random_state=42)
X_train.shape, x_test.shape, Y_train.shape, y_test.shape
```

```python
scaler = MinMaxScaler()

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# now convert into dataframe also add all three couumns
X_train_scaled = pd.DataFrame(X_train_scaled, columns=['fixed acidity', 'volatile acidity', 'free sulfur dioxide'])
x_test_scaled = pd.DataFrame(x_test_scaled, columns=['fixed acidity', 'volatile acidity', 'free sulfur dioxide'])
X_train_scaled.head()
```
