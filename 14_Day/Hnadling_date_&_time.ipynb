# HANDLING DATE AND TIME

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime
```

```python
# import both dataset
df1 = pd.read_csv('date.csv')
df2 = pd.read_csv('time.csv')
```

```python
df1.head()
```

```python
# convert date colum into date_time
df1['Date'] = pd.to_datetime(df1['Date'])
```

```python
df1['date_year'] = df1['Date'].dt.year
# now for month
df1['date_month_no'] = df1['Date'].dt.month
# now for day
df1['date_day_no'] = df1['Date'].dt.day
```

```python
df1.head()
```

```python
# now day of week 
df1['date_day_name'] = df1['Date'].dt.day_name()

# now for week day no
df1['date_week_day_no'] = df1['Date'].dt.weekday
df1['date_week_no'] = df1['Date'].dt.isocalendar().week
```

```python
df1.head()
```

```python
df1['date_is_weekand'] = np.where(df1['date_day_name'].isin(['Saturday', 'Sunday']), 1, 0)
```
