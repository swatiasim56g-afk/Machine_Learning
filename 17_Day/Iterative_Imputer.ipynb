```python
import pandas as pd

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Numerical columns
num_cols = ["Age", "Salary", "Experience"]

# Create Iterative Imputer
imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

# Apply imputer
df[num_cols] = imputer.fit_transform(df[num_cols])

# Check missing values using loop
for col in num_cols:
    print(f"{col}: {df[col].isnull().sum()} missing values")
```

## See Before and After Using a Loop

```python
# Before imputation
print("Before Imputation:")

for col in num_cols:
    print(f"{col}: {df[col].isnull().sum()} missing values")


# Create imputer
imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

# Fill missing values
df[num_cols] = imputer.fit_transform(df[num_cols])


# After imputation
print("\nAfter Imputation:")

for col in num_cols:
    print(f"{col}: {df[col].isnull().sum()} missing values")
```
