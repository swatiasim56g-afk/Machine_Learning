# Z Score Outlier Detection

## Z-Score using formula

```python
import pandas as pd

df = pd.DataFrame({
    "Salary": [30000, 35000, 40000, 45000, 50000, 200000]
})

# Mean
mean = df["Salary"].mean()

# Standard deviation
std = df["Salary"].std()

# Z-score using formula
df["z_score"] = (df["Salary"] - mean) / std

print(df)
```

```python
outliers = df[abs(df["z_score"]) > 3]

print(outliers)
```

```python
abs(df["z_score"]) > 3
```

## Z-Score Using Mean ± 3 Standard Deviations

```python
mean = df["Salary"].mean()
std = df["Salary"].std()

lower_limit = mean - 3 * std
upper_limit = mean + 3 * std

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

print(outliers)
```

## Z-Score Capping

```python
mean = df["Salary"].mean()
std = df["Salary"].std()

lower_limit = mean - 3 * std
upper_limit = mean + 3 * std

df["Salary_capped"] = df["Salary"].clip(
    lower=lower_limit,
    upper=upper_limit
)

print(df)
```
