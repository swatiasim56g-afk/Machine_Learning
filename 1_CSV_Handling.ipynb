## 4 : Index_Col Parameter
some time in the data we have two id columa like 1,2,3... and the data have also own id 123, 344 , 434 etc... so we will remove one id colum then we used index_col parameter when we put the colum name then it stayed only that colum in the data 

```python
pd.read_csv('data.csv',index_col='enrolled_Id')
```

## 5 : Header Parameter

```python
pd.read_csv('data.csv',header=1)
```

## 6 : Use_Col Parameter 
only used those colum which we used for that we used colum

```python
pd.read_csv('data.csv',usecols=['enrolled_Id','first_name','last_name','email'])
```

## 9 : Dtype Parameter 
sometime we have the column of float we want to convert into int so then we used dtype parmater to convert it.


```python
# convert column to integer
pd.read_csv('data.csv',dtype={'enrolled_Id':int})
```

## 10 : Handling Dates

```python
# Handling dates

pd.read_csv('data.csv',parse_dates=['enrolled_date'])
```
