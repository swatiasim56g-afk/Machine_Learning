```python
import numpy as np
import matplotlib.pyplot as plt

# Create the same number of data points for every dimension
n = 100

```

```python
# -----------------------------
# 1 DIMENSION
# -----------------------------

data_1d = np.random.randn(n)

plt.figure(figsize=(10, 2))
plt.scatter(data_1d, np.zeros(n))
plt.xlabel("Dimension 1")
plt.title("1 Dimensional Space")
plt.yticks([])
plt.show()


```

```python
# -----------------------------
# 2 DIMENSIONS
# -----------------------------

data_2d = np.random.randn(n, 2)

plt.figure(figsize=(6, 5))
plt.scatter(data_2d[:, 0], data_2d[:, 1])
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.title("2 Dimensional Space")
plt.show()


```

```python
# -----------------------------
# 3 DIMENSIONS
# -----------------------------

data_3d = np.random.randn(n, 3)

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    data_3d[:, 0],
    data_3d[:, 1],
    data_3d[:, 2]
)

ax.set_xlabel("Dimension 1")
ax.set_ylabel("Dimension 2")
ax.set_zlabel("Dimension 3")
ax.set_title("3 Dimensional Space")

plt.show()


```

```python
# -----------------------------
# 4 DIMENSIONS
# -----------------------------

data_4d = np.random.randn(n, 4)

plt.figure(figsize=(6, 5))

scatter = plt.scatter(
    data_4d[:, 0],
    data_4d[:, 1],
    c=data_4d[:, 2],
    s=50 + (data_4d[:, 3] + 3) * 20
)

plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.title("4 Dimensional Space")
plt.colorbar(scatter, label="Dimension 3")

plt.show()


```

```python
# -----------------------------
# 5 DIMENSIONS
# -----------------------------

data_5d = np.random.randn(n, 5)

plt.figure(figsize=(7, 6))

scatter = plt.scatter(
    data_5d[:, 0],
    data_5d[:, 1],
    c=data_5d[:, 2],
    s=50 + (data_5d[:, 3] + 3) * 20,
    alpha=0.7
)

plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.title("5 Dimensional Space")
plt.colorbar(scatter, label="Dimension 3")

plt.show()


```
